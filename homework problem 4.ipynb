```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

# Question 3 part B

```python
data=[5,7,7,8,10]
def kde(x_val, x,h):
    X=np.array(x)
    guass=lambda z: np.exp(-z**2/2)/np.sqrt(2*np.pi)
    K=guass((x_val-X)/h)/h
    return K.mean()


test_points=[6, 6.5, 7, 8, 8.5]
y_val=[]
for point in test_points:
    y_val.append(kde(point, data, 1.5))

y_val
```

# Question 4

```python
#Load data
heart=pd.read_csv('./data/heart_failure_clinical_records_dataset.csv')
heart.head()
```

## 4A

```python
sns.kdeplot(data=heart['ejection_fraction'], fill=True, color="blue")
plt.title("Univariate KDE Plot")
plt.show()
```

## 4B

```python
for i in range(15):
    data=heart.sample(frac=1.0, replace=True)
    sns.kdeplot(data=data['ejection_fraction'], fill=False)
plt.title("15 kde plots")
plt.show()
```

## 4C

The first plot will be reliable on the ranges where the 15 plots all overlap. For example, in the range from 0 to 30, all of the 15 plots overlap with each other which indicates that the true density is very close to our predicted density in that range. However, at 30, 40, and 60, there are large variations in the overlapping KDEs. This indicates that at and around these values, our prediction may be less reliable. The plots are struggling to agree on what the true density is.
